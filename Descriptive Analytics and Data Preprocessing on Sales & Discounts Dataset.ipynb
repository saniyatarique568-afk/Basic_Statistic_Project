```python
#plotting histogram for all numeric columns
numeric_df.hist(figsize=(15, 10), bins=15, edgecolor="black")

plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
#detailed histogram for each numeric column
for column in numeric_df.columns:
    plt.figure(figsize=(8, 4))
    plt.hist(numeric_df[column], bins=15, edgecolor="black")
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
brand_sales = df.groupby("Brand")["Total Sales Value"].sum()
```

```python
brand_sales.plot(kind="bar", figsize=(10, 5))

plt.title("Total Sales by Brand")
plt.xlabel("Brand")
plt.ylabel("Total Sales Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show() #plotting a bar chart for total sales by brand
```

```python
#frequency of each brand
brand_counts = df["Brand"].value_counts()

brand_counts.plot(kind="bar", figsize=(10, 5))

plt.title("Frequency of Each Brand")
plt.xlabel("Brand")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#skewness of numeric columns
print(numeric_df.skew())
```

```python
#creating a box plot to identify potential outliers
for column in numeric_df.columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=numeric_df[column])

    plt.title(f"Box Plot of {column}")
    plt.xlabel(column)
    plt.tight_layout()
    plt.show()
```

```python
for column in numeric_df.columns:
    Q1 = numeric_df[column].quantile(0.25)
    Q3 = numeric_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = numeric_df[
        (numeric_df[column] < lower_limit) |
        (numeric_df[column] > upper_limit)
    ][column]

    print(column, ":", len(outliers), "potential outliers")
```
